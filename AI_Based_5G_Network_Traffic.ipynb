```python
import numpy as np
import pandas as pd

import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score
from sklearn.metrics import classification_report

import joblib

print("All libraries imported successfully!")
print("Welcome to AI Based 5G Network Traffic Project!")
```

```python

# STEP 3: CREATE 5G NETWORK DATASET

import numpy as np
import pandas as pd

# Generate sample network data
np.random.seed(42)

n = 2000

data = pd.DataFrame({

    "connected_users":
    np.random.randint(10, 2000, n),

    "bandwidth_mbps":
    np.random.randint(50, 1000, n),

    "signal_strength":
    np.random.randint(1, 101, n),

    "packet_rate":
    np.random.randint(100, 10000, n),

    "hour":
    np.random.randint(0, 24, n)

})

# Display dataset
print("5G NETWORK DATASET CREATED!")
print("\nFirst 10 rows of dataset:")

display(data.head(10))

print("\nDataset Shape:", data.shape)

print("\nDataset Information:")
print(data.info())
```

```python

# STEP 4: TRAFFIC CLASSIFICATION

# Calculate network traffic score

traffic_score = (
    data["connected_users"] / 2000 * 0.35 +
    data["packet_rate"] / 10000 * 0.35 +
    (1 - data["bandwidth_mbps"] / 1000) * 0.20 +
    data["hour"].apply(
        lambda x: 0.10 if 18 <= x <= 22 else 0.03
    )
)

# Assign traffic categories

data["traffic_level"] = pd.cut(
    traffic_score,
    bins=[-np.inf, 0.30, 0.55, np.inf],
    labels=["Low", "Medium", "High"]
)

# Display classified dataset

print("TRAFFIC CLASSIFICATION COMPLETED!")

display(data.head(15))

print("\nTraffic Category Distribution:")
print(data["traffic_level"].value_counts())
```

```python

# STEP 5: TRAIN AI MODEL

from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier

# Separate input and output
X = data.drop("traffic_level", axis=1)
y = data["traffic_level"]

# Split dataset into training and testing
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training data:", X_train.shape)
print("Testing data:", X_test.shape)

# Create Random Forest model
model = RandomForestClassifier(
    n_estimators=100,
    random_state=42
)

# Train AI model
model.fit(X_train, y_train)

print("\n================================")
print("AI MODEL TRAINING COMPLETED!")
print("================================")

print("Algorithm: Random Forest")
print("Training Samples:", len(X_train))
print("Testing Samples:", len(X_test))
```

```python

# STEP 6: MODEL EVALUATION

from sklearn.metrics import accuracy_score
from sklearn.metrics import classification_report
from sklearn.metrics import confusion_matrix

# Predict testing data
y_pred = model.predict(X_test)

# Calculate accuracy
accuracy = accuracy_score(y_test, y_pred)

print("================================")
print("AI MODEL PERFORMANCE")
print("================================")

print("\nModel Accuracy:",
      round(accuracy * 100, 2), "%")

print("\nClassification Report:")
print(classification_report(y_test, y_pred))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred))
```

```python

# STEP 7: NEW 5G NETWORK TRAFFIC PREDICTION

import pandas as pd

# New network input
new_network = pd.DataFrame([{
    "connected_users": 1500,
    "bandwidth_mbps": 200,
    "signal_strength": 80,
    "packet_rate": 8500,
    "hour": 20
}])

# Predict traffic
prediction = model.predict(new_network)

print("====================================")
print("5G NETWORK TRAFFIC PREDICTION")
print("====================================")

print("Connected Users: 1500")
print("Bandwidth: 200 Mbps")
print("Signal Strength: 80")
print("Packet Rate: 8500")
print("Hour: 20")

print("\nPredicted Traffic:", prediction[0])

print("\nAI PREDICTION COMPLETED!")
```

```python

# STEP 8: SAVE AI MODEL AND DATASET

import joblib

# Save trained AI model
joblib.dump(model, "5G_Traffic_Model.pkl")

# Save dataset
data.to_csv("5G_Network_Traffic_Dataset.csv", index=False)

print("================================")
print("PROJECT FILES SAVED SUCCESSFULLY")
print("================================")

print("1. 5G_Traffic_Model.pkl")
print("2. 5G_Network_Traffic_Dataset.csv")
print("3. AI Model Accuracy:", round(accuracy * 100, 2), "%")
```
